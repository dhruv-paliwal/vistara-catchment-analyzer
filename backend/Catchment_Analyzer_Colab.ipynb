```python
# Set up FastAPI server script
server_code = '''
from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from typing import List
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.ensemble import RandomForestRegressor
import math

app = FastAPI(title="Catchment Analyzer ML API")
app.add_middleware(CORSMiddleware, allow_origins=["*"], allow_methods=["*"], allow_headers=["*"])

EARTH_RADIUS_KM = 6371.0

class CatchmentPointItem(BaseModel):
    id: str
    name: str
    lat: float
    lng: float
    type: str
    dailyTraffic: float
    populationDensity: float

class ClusterRequest(BaseModel):
    points: List[CatchmentPointItem]
    epsKm: float = 1.8
    minSamples: int = 3

@app.get("/api/health")
def health():
    return {"status": "healthy", "source": "Google Colab Microservice"}
'''

with open('main.py', 'w') as f:
    f.write(server_code)
print("Server script written successfully!")
```

```python
import nest_asyncio
import uvicorn
from pyngrok import ngrok

# Enter your free ngrok auth token if required: ngrok.set_auth_token("YOUR_TOKEN")
public_url = ngrok.connect(8000).public_url
print(f"\n🔥 Public API URL for your Next.js App: {public_url}\n")
nest_asyncio.apply()
uvicorn.run('main:app', host='0.0.0.0', port=8000)
```
